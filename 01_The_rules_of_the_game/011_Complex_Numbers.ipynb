# The number $i$


Quantum mechanics embodies a rather intriguing relationship with complex numbers. To introduce them, we will start by recalling that a real number, squared, *always* gives a positive result
$$
a^2 >0
$$  

for example $~~2^2 =  4~~$ but also $~~(-2)^2 = 4~~$. Therefore, the square root of a real number <u>*only* exists if that number is positive</u>.  
For example $\sqrt{4}=\pm 2$, while for a negative number, we don't know what it means to have

$$\sqrt{-4}= \, ?$$


**Question**: $~~$ How could we define the <u>square root of a negative real number</u>?

**Answer**: to do this, it is necessary to <u>expand the set of real numbers</u>. 


<div class="alert alert-block alert-info",text-align:center>
<p style="text-align: left; color: navy"> 

<b>Definition:</b> <i>(number $i$) </i>
<br>
        the existence of a <i>new number</i>, $~i$, is postulated, as the unique solution of the equation 
<br><br>
$$ i^2 = -1$$
</p>
</div>

Equivalently, we could have required that $~i = \sqrt{-1}$ .

This given, we can now define the square root of any negative number. For example $-4$

$$
(2i)(2i) = 4 i^2 = - 4
$$

so $2i$ will be the searched for square root.  With the number $i$, operations are performed just like with real numbers


$$ i + i = 2i $$

$$ i - i = 0 $$

$$ i + 2i = 3 i$$

$$i^3  = i*i*i = i^2 * i = -i$$

$$\frac{i}{i} = 1$$

Note that the *multiplicative inverse* $1/i$ is also $-i$

\begin{align}
\frac{i}{i}  &=   1 \nonumber\\ \rule{0mm}{8mm}
i (-i) &= - i^2  =  1
\end{align}

therefore there is an important identification

$$ -i = \frac{1}{i} $$

<b>Summary</b>: the solution to the problem posed consists of <i>extending</i> the body of real numbers ${\mathbb R}$ to that of complex numbers ${\mathbb C}$, which now include the number $i$

## Cartesian Form





A *complex number*, $z \in {\mathbb C}$, is represented in *Cartesian form* by <u>two 
real numbers $x,y\in {\mathbb R}$</u> 
<br>

$$
z = x + i  y   ~~~~ \hbox{where}~~~ 
\left\{\begin{array}{cl} x &\hbox{is the }   real~ part\\
y & \hbox{is the } imaginary~part
\end{array}
\right.
$$


A complex number can be represented in a plane:  the **complex plane**. The real part on the *horizontal axis*, and the imaginary part on the *vertical axis*

## Polar Form




<div class="alert alert-block alert-info",text-align:center>
<p style="text-align: center;"> <p style="text-align: left ;color: navy;">  

<b> Theorem:</b> (Euler's Formula).
 Given an angle $\theta \in (0,2\pi)$ the following two expressions are equivalent
<br>
<br>
$$
\cos\theta + i\sin \theta = e^{i\theta} 
$$
</div>
<details>
<summary><p style="text-align:right ; color:grey"> >>Proof
</p></summary>
<br>
The proof of Euler's Formula comes from expanding both sides into Taylor series around $\theta = 0$ and verifying that both series are equal
<br> 
\begin{array}{rcl}
e^{i\theta} &= 1 + \i\theta + \frac{1}{2}(\i\theta)^2 + \frac{1}{3!}(\i\theta)^3+\, ... \\
    &= 1 -\frac{1}{2}\theta^2 ~+~ ... ~+~ \i \left(\theta - \frac{1}{3!} \theta^3+ \, ...\right) \\
    &= \cos \theta  + i \sin \theta 
\end{array} 
</details>

The number $z=x + i y$ can be represented in *polar form*
<br> 

$$
z = \rho e^{i\theta} = \rho (\cos\theta + i \sin\theta)  
$$

from where we obtain the Cartesian components

$$x=\rho\cos\theta ~~,~~y=\rho\sin\theta$$

and therefore $\rho^2 = x^2 + y^2$

The real numbers $\rho$ and $\theta$ are called *modulus* and *phase*.
<br>

The phases $\theta$ and $\theta+ 2\pi$ represent the *same* complex number
<br>

$$
z = \rho e^{i\theta} = \rho e^{i(\theta + 2\pi)} 
$$


This is because the functions $\cos \theta$ and $\sin\theta$ are periodic

$$
\sin\theta  = \sin(\theta + 2\pi)~~~~~~~\cos\theta  = \cos(\theta +2\pi)
$$

  In polar form, the *complex conjugate* is obtained by *changing the sign* of the phase
<br>

$$
z = \rho e^{i\theta} ~\rightarrow ~z^* = \rho e^{-i\theta} = \rho \cos\theta - i \rho \sin\theta \hspace{4cm}
$$

````python
'''Forma polar'''
r=2
th=3.5
z = r*np.exp(th*1j)
print(z)
x=z.real
y=z.imag
print('z=r exp(i th)=',np.round(x,3) + np.round(y,3)*1j)


''' Dibujamos en el plano complejo '''
tQ.plot_2D_plane(left=-int(abs(x))-1,right=int(abs(x))+1,up=int(abs(y))+1,down=-int(abs(y))-1)
tQ.draw_vector(z.real,z.imag)
tQ.draw_vector(z.conjugate().real,z.conjugate().imag,vcolor='r')
````

i &= e^{i\pi/2} \\ \rule{0mm}{8mm}
-1 &= e^{i\pi} \\ \rule{0mm}{8mm}
i &= e^{3i\pi/2} =e^{-i\pi/2} \\  \rule{0mm}{8mm}
i^i &= (e^{i\pi/2})^i = e^{i^2 \pi/2} = e^{-\pi/2}  \\  \rule{0mm}{8mm}
i^{2+i} &= (e^{i\pi/2})^{(2+i)}= e^{i\pi/2(2+i)} = e^{i\pi} e^{-\pi/2} = -e^{-\pi/2}  \rule{0mm}{8mm}
\end{align}

## Conversion between Cartesian and Polar Forms


The conversion from *polar to Cartesian* representation is very simple
thanks to Euler's formula


<br>

$$
z = r e^{i\theta} = x + i y ~~~\hbox{ with }  ~~~\left\{\begin{array}{l} x=r \cos \theta \\ \rule{0mm}{4mm} y = r\sin \theta
\end{array} \right.
$$

<br>

The inverse conversion, *from Cartesian to polar* is a bit more delicate. Formally it would be

<br>

$$
z = x + i y  = r e^{i\theta} ~~~ \hbox{ with } ~~~ \left\{\begin{array}{l} r=\sqrt{x^2+y^2} \\  \rule{0mm}{4mm} \theta = \arctan(y/x)
\end{array} \right.
$$


When it comes to the moment of truth, the sign of the function $\arctan(y/x)$ must be determined. The following function examines this by looking at the signs of $x$ and $y$ to know in which quadrant we are.

````python
'Conversión of cartesian into polar'
def cartes2polar(z):

    r = np.abs(z)    
    y = z.imag
    x = z.real
    
    if r==0:
        print('the number 0+i0 does not admit a polar representation')
        th='indefinite'
    elif x==0 and y>0: 
        th=np.pi/2
    elif x==0 and y<0:
        th=3*np.pi/2
    elif x>0 and y>=0:
        th=np.arctan(y/x)
    elif x<0 and y>=0:
        th=np.arctan(-y/x)+np.pi/2
    elif x<0 and y<0:
        th=np.arctan(y/x)+np.pi
    elif x>0 and y<0:
         th=np.arctan(-y/x)+3*np.pi/2.       
            
    return r,th

#el signo correcto también se puede conseguir usando la funcion np.arctan2(x,y)
````

# Basic Operations

The complex numbers ${\mathbb C}$ form a mathematical structure called a *field*. This means they allow two *internal* operations: **addition** and **multiplication**. We will study them separately.

## Complex Conjugation

Every complex number, $z$, has an *associated* number, $z^*$, called the *complex conjugate*, which is obtained by changing $i \to -i$
<br>

$$
z = x+ i y ~~~~\leftrightarrow~~~~ z^* = x - i y \hspace{2cm}
$$
<br>

It is evident that $(z^*)^* = z$, it is an *involution*.

````python
'''Complex comjugation'''
zc = z.conjugate()
print('z*=(x+iy)*=',z.real +1j*zc.imag)

''' Representation on the complex plane'''
tQ.plot_2D_plane(left=-int(abs(x))-1,right=int(abs(x))+1,up=int(abs(y))+1,down=-int(abs(y))-1)
tQ.draw_vector(z.real,z.imag,'b')
tQ.draw_vector(zc.real,zc.imag,vcolor='r')

````

## Addition

In *Cartesian* representation, the *real and imaginary parts are added separately*

<br>

$$
(a + i b) + (c + i d) = (a+c) + i (b+d)\hspace{6cm}
$$ 


<br>



Subtraction is obvious, since $a,b,c,d$ can be negative numbers. 


````python
'''Adding in cartesian'''

z1 = 1+4j
z2 = 2+2j

'''Adding and subtracting'''
zs = z1+z2
zd = z1-z2

print('z1+z2=',zs)
print('z1-z2=',zd)
print('************************************')



tQ.plot_2D_plane(left=-2,right=4,up=7,down=-1) #cambiar las dimensiones para que encuadrar la figura
tQ.draw_vector(z1.real,z1.imag,'b')
tQ.draw_vector(z2.real,z2.imag,'b')
tQ.draw_vector(zs.real,zs.imag,vcolor='r')
tQ.draw_vector(zd.real,zd.imag,vcolor='g')
````

z + w = \rho e^{i\theta} + \sigma e^{i\phi} = (\rho\cos\theta + \sigma\cos\phi) + i(\rho\sin\theta +  \sigma\sin\phi) \hspace{6cm}
$$

## Multiplication

In *Cartesian form*, multiplication is complicated, as all factors must be multiplied with each other, and considering that $i^2= -1$

$$
(a + i b) (c + i d)  =ac +  ai d +i bc +i^2 bd = (ab - bd) + i(ac + bd)\hspace{6cm}
$$ 

To find the product of two complex numbers $z=r e^{i\theta}$ and $w=s e^{i\phi}$ written in polar form, the moduli are multiplied and the phases are added

$$
z w = r e^{i\theta} s e^{i\phi} = rs\,   e^{i(\theta + \phi)} 
$$

````python
z1 = 3*np.exp(1j*0.5)
z2 = 1*np.exp(-1j*0.7)
'''Product'''
print('z1*z2 = ', np.round(z1*z2,4))
print('z1**3 = ', np.round(pow(z1,6)))
print('************************************')
````

## Absolute value

The square of a real number $a\in {\mathbb R}$ is another positive real number $a^2 >0$. This allows us to define
the absolute value $|a| = \sqrt{a^2}$ which is the same for $a$ and for $-a$.

This does not happen with a complex number $z$. $~$ In fact,  

$$z^2 = x^2 - y^2 +2 i xy $$ 

is complex.$~$ However, the product of a number by its conjugate is a *real* and *positive* number

$$
z z^*  = (x + i y) (x- i y) = x^2 + y^2 >0
$$


This allows us to define the *absolute value* of a complex number

$$
|z| = \sqrt{z z^*} = \sqrt{x^2 + y^2}
$$


The *absolute value* of a phase is 1

$$
|e^{i\theta}| = \sqrt{ e^{i\theta}   e^{-i\theta}}=\sqrt{ e^{i(\theta-\theta)}}=\sqrt{e^0} = 1
$$


The *absolute value* of a complex number with the *modulus* written in polar form

$$
|z| = \sqrt{zz^*} = \sqrt{\rho e^{i\theta} \rho e^{-i\theta}}=\sqrt{\rho^2} = \rho
$$

## Division

Like multiplication, in Cartesian form, division **is not simple**. Let $z = a+ i b$ and $w=c+i d$ 
<br>
<br>

$$
\frac{z}{w} = \frac{z}{w}\frac{w^*}{w^*} = \frac{( a+  i b)(c- i d)}{|w|^2} = \frac{ac+bd +  i(bc-ad)}{c^2+d^2}  = \frac{ac+bd}{c^2+d^2} + i\frac{bc-ad}{c^2+d^2}
$$


In polar form, division is as simple as multiplication. You take the quotient of the magnitudes and the difference of the phases

$$
\frac{z}{w} = \frac{\rho e^{i\theta}}{\sigma e^{i\phi}} = \frac{\rho}{\sigma}
e^{i(\theta-\phi)}
$$

````python
'''Absolute value'''
print('|z1|=',abs(z1))
print('check |z1|=',np.sqrt(z1*z1.conjugate()).real) 
print('************************************')


'''Division'''
print('z1/z2=',np.round(z1/z2,5))
print('check z1/z2=', np.round(z1*z2.conjugate()/(z2*z2.conjugate()),5))
````

## Null sums

On many occasions, we will encounter the following representation of the number zero (complex) $0 = 0 + i 0$

$$
\sum_{k=0}^{N-1} e^{2\pi i k/N} =   e^{2\pi i\, 0/N} +  e^{2\pi i\, 1/N}  +~...~ +   e^{2\pi i\, (N-2)/N}+   e^{2\pi i\, (N-1)/N} ~=~  ~0
 $$

To convince yourself that this identity is true, we will represent the complex numbers and their sum. You can change $N$ and also multiply by a constant modulus

````python
# change the number N
N=9
rho=1

''' define the phases'''
lista_de_fases=np.exp(2*np.pi*1j*np.array(range(N))/N)
#print('lista de fases =', np.round(lista_de_fases,2))


''' plot the complex numbers '''
tQ.plot_2D_plane(fsize=(6,6))
for vec in rho*lista_de_fases:
    tQ.draw_vector(x=vec.real,y=vec.imag)

#draw_unit_circle()
plt.gca().add_patch(plt.Circle((0.,0.),1.,color='black',fill=False)) 

''' compute the sum '''
#print(lista_de_fases)
print(np.round(sum(rho*lista_de_fases),10))
````

Let $1\leq j \leq N-1$ be an integer by which we multiply all the phases. The result is the same

$$
\sum_{k=0}^{N-1} e^{2\pi \i j k/N} =   e^{2\pi \i\, 0/N} +  e^{2\pi \i\, j/N}  +~...~ +   e^{2\pi \i\, j(N-2)/N}+   e^{2\pi \i\, j(N-1)/N} ~=~  ~0
 $$


````python
# change the number N
N=4
rho=1
j=1
''' define the phases'''
lista_de_fases=np.exp(2*j*np.pi*1j*np.array(range(N))/N)
#print('lista de fases =', np.round(lista_de_fases,2))


''' plot the complex numbers '''
tQ.plot_2D_plane(fsize=(6,6))
for vec in rho*lista_de_fases:
    tQ.draw_vector(x=vec.real,y=vec.imag)

    
#draw_unit_circle()
plt.gca().add_patch(plt.Circle((0.,0.),1.,color='black',fill=False)) 

''' compute the sum '''
#print(lista_de_fases)
print(np.round(sum(rho*lista_de_fases),10))
````

However, if $j = 0, N, 2N,... = 0\,\hbox{mod} N$, then the sum does not cancel out and its value is equal to $N$. 

Let's take for example $j=N$ 

$$
\sum_{k=0}^{N-1} e^{2\pi  i (3N) k/N} = \sum_{k=0}^{N-1} e^{2\pi  i  k}  =  \sum_{k=0}^{N-1} 1 =~  ~N
 $$


````python
# change the number N
N=9
rho=1
j=N
''' define the phases'''
lista_de_fases=np.exp(2*j*np.pi*1j*np.array(range(N))/N)
#print('lista de fases =', np.round(lista_de_fases,2))


''' plot the complex numbers '''
tQ.plot_2D_plane(fsize=(6,6))
for vec in rho*lista_de_fases:
    tQ.draw_vector(x=vec.real,y=vec.imag)

#draw_unit_circle()
plt.gca().add_patch(plt.Circle((0.,0.),1.,color='black',fill=False)) 

''' compute the sum '''
#print(lista_de_fases)
print(np.round(sum(rho*lista_de_fases),10))
````

### Norm
<a id="norm"></a>

The fact that, for any complex numbers $a,b\in \mathbb{C}$

- Homogeneity: $|\lambda a| = \lambda |a| $

- Non-degeneracy: $|a| = 0 \Leftrightarrow a = 0$. 

- Triangular inequality: $ |a + b| \leq |a| + |b|$. 

makes the **absolute value** into a **norm**, and the vector space of complex numbers into a **normed vector space**.
